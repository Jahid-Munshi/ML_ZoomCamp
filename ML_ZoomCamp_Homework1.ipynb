### Question 3. Fuel types (1 point)

```python
fuel_types_count = df["fuel_type"].nunique()
print("Q3. Fuel types:", fuel_types_count)
```

### Question 4. Missing Values (1 point)

```python
missing_values = (df.isnull().sum() > 0).sum()
print("Q4. Missing values:", missing_values)
```

### Question 5. Max fuel efficiency (1 point)

```python
max_fuel_eff = df["fuel_efficiency_mpg"].max()
print("Q5. Max fuel efficiency:", max_fuel_eff)
```

### Question 6. Median value of horsepower (1 point)

```python
median_hp_earliest = df[df["model_year"] == df["model_year"].min()]["horsepower"].median()
median_hp_latest = df[df["model_year"] == df["model_year"].max()]["horsepower"].median()

if median_hp_latest > median_hp_earliest:
    hp_change = "Yes, it increased"
elif median_hp_latest < median_hp_earliest:
    hp_change = "Yes, it decreased"
else:
    hp_change = "No"

print("Q6. Median value of horsepower:", hp_change)
```

### Question 7. Sum of weights (1 point)

```python
sum_weights = df["vehicle_weight"].sum()
print("Q7. Sum of weights:", sum_weights)

scaled_sum = sum_weights / 1e6
print("Scaled sum (in millions):", scaled_sum)
```
